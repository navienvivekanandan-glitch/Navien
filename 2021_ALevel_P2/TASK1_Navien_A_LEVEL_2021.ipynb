```python
#Task 1.1

def task1_1(input_value):
    if str(input_value).isdigit() != True:
        return -1
    elif len(str(input_value)) != 14:
        return -1
    else:
        value = 0
        input_value = str(input_value)
        for i in range(len(input_value)):
            if i % 2 == 1: 
                mid_value = int(input_value[i]) * 2
                if mid_value > 9:
                    mid_value = str(mid_value)
                    num1 = mid_value[:1]
                    num2 = mid_value[1:]
                    value += int(num1)
                    value += int(num2)
                else:
                    value += int(mid_value)
                 
            else:
                value += int(input_value[i])
            
        
        check_digit = 0
        while value % 10 != 0:
            value += 1
            check_digit += 1
        
        return check_digit
    
print(task1_1(14576567654934))
```

```python
#Task 1.2

#Test 1
print(task1_1(56565656565656))
    #Valid -> returns check digit = 4
    
#Test 2

print(task1_1('56565656565656'))
    #Valid -> returns check digit = 4
    
#Test 3
print(task1_1('56565656A!5656'))
    #Invalid, is not purely just numbers -> returns -1 
    
#Test 4
print(task1_1('565656565656'))
    #Invalid, length is shorter -> returns -1 
```

```python
#Task 1.3

def task1_3(input_value):
    if str(input_value).isdigit() == False:
        return False 
    if len(str(input_value)) != 15:
        return False
    
    input_value = str(input_value)
    new_str = input_value[:14]
    check_digit = input_value[14:]
    calc_check_digit = task1_1(new_str)   
    
    if str(calc_check_digit) == check_digit:
        return True
    else:
        return False

#Test 1
print(task1_3(565656565656564))
    #Valid -> returns check digit = 4
    
#Test 2

print(task1_3('565656565656564'))
    #Valid -> returns check digit = 4
    
#Test 3
print(not task1_3('56565656A!5656'))
    #Invalid, is not purely just numbers -> returns True (not False)
    
#Test 4
print(not task1_3('565656565656'))
    #Invalid, length is shorter -> returns True (not False)
```
