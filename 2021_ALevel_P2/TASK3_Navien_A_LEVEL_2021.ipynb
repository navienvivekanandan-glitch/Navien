```python
#Task 3.2

class SortedLinkedList(LinkedList):
    def __init__(self):
        super().__init__()
    
    def insert(self,integer_value):
        node = Node(integer_value)
        if self.head == None or integer_value < self.head.data:
            node.next = self.head
            self.head = node
        else:
            curr = self.head
            while curr.next != None and curr.next.data <= integer_value :
                curr = curr.next
            
            node.next = curr.next
            curr.next = node
```

```python
#Task 3.3

class Queue(LinkedList):
    def __init__(self):
        self.front = None
        self.back = None
        super().__init__()
    
    def enqueue(self,integer_value):
        node = Node(integer_value)
        
        if self.head == None:
            self.head = node
            self.front = node
            self.back  = node
        else:
            self.back.next = node
            self.back = node
    
    def dequeue(self):
        if self.head == None:
            return None
        else:
            value = self.head.data
            self.head = self.head.next
        
            if self.head == None:
                self.back = None
                self.front = None
        
            return value
```
