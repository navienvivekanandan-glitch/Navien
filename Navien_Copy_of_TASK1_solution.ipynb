```python
# Task 1.1
import pymongo,json
client = pymongo.MongoClient('127.0.0.1',27017)
db = client.get_database('ParcelsDB')
coll = db.get_collection('Parcels')
```

```python
# Task 1.2
def loadparcels():
    file = open('TASK1_2.JSON','r')
    data = json.load(file)
    
    coll.delete_many({})
    if len(data) != None:
        coll.insert_many(data)
        print("Parcel record loaded successfully")
    else:
        print("No parcel record found in JSON file")
    file.close()

    
```

```python
# Task 1.3
def displayParcel(parcelID):
    result = coll.find_one({"parcelID":parcelID})
    if result != None:
        print("--- Parcel Found ---")
        print("Parcel ID: ",result["parcelID"])
        print("Recipient : ",result["recipient"])
        print("Priority : ",result["priority"])
        print("Status: ",result["status"])
    
    else:
        print("--- Parcel Not Found ---")
```

```python
# Task 1.4
def priorityParcels(priorityType):
    results = coll.find({"priority": priorityType})
    count = 0
    print(f"{'parcel ID':<12}{'recipient':<20}")
    for parcel in results:
        print(f"{parcel['parcelID']:<12}{parcel['recipient']:<20}")
        count += 1
    print("Total no. of parcels: ",count)
```

```python
# Main

loadparcels()
print()

displayParcel("P1001")


priorityParcels("Standard")
```
