```python
#Task 3

#Task 3.1

def heading_html(text):
    heading = '<h1>'+text+'</h1>'
    return heading

def subheading_html(text):
    heading = '<h2>'+text+'</h2>'
    return heading

def paragraph_html(text):
    heading = '<p>'+text+'</p>'
    return heading

print(heading_html("My first page"))
print(subheading_html("My first page"))
print(paragraph_html("My first page"))
```

```python

#Task 3.2

def hyperlink_html(text):
    url,string = text.split(',')

    hyperlink = "<a href=" +'"'+ url + '"' + ">" + string + "</a>"
    return hyperlink

print(hyperlink_html("https://www.cambridgeinternational.org,Click here"))
```

```python

#Task 3.3

def body(lst):
    formatted_string = ''
    for line in lst:
        if line.startswith('heading:') == True:
            line = line.split('heading:')
            element = line[0]
            text = line[1]
            formatted_string += heading_html(text)
        elif line.startswith('subheading:') == True:
            line = line.split('subheading:')
            element = line[0]
            text = line[1]
            formatted_string += subheading_html(text)
        elif line.startswith('hyperlink:') == True:
            line = line.split('hyperlink:')
            element = line[0]
            text = line[1]
            formatted_string += hyperlink_html(text)
        elif line.startswith('paragraph:') == True:
            line = line.split('paragraph:')
            element = line[0]
            text = line[1]
            formatted_string += paragraph_html(text)
        else:
            return False
            
    formatted_string = '<html><body>' + formatted_string + '</html></body>'

    return formatted_string

def read_data(filename):
    file = open(filename,'r')
    lst = []
    for line in file:
        line = line.strip()
        lst.append(line)
    formatted_string = body(lst)
    file.close()
    return formatted_string


def write_data(filename,formatted_string):
    filename = filename + ".html"
    file = open(filename,'w')
    file.write('<!doctype html>')
    file.write(formatted_string)
    file.close()
    return

formatted_string = read_data('firstPage.txt')
write_data('test',formatted_string)

formatted_string = read_data('website.txt')
write_data('theWebsite',formatted_string)

```
