```python
#Task 1 

#Task 1.1
import random

hand1 = []
hand2 = []
colour_lst = ['green','red','blue']
for i in range(10):
    colour = colour_lst[random.randint(0,2)]
    number = random.randint(1,4)
    hand1.append([colour,number])
    colour = colour_lst[random.randint(0,2)]
    number = random.randint(1,4)
    hand2.append([colour,number])
    
print(hand1)
print(hand2)
```

```python
#Task 1.2

def compare_colours(colour1,colour2):
    if colour1 == colour2:
        return 3
    if colour1 == 'red' and colour2 == 'green':
        return 1 
    if colour1 == 'green' and colour2 == 'blue':
        return 1
    if colour1 == 'blue' and colour2 == 'red':
        return 1
    return 2 
```

```python
#Task 1.3
def sort(hand):
    for j in range(len(hand)-2):
        for i in range(len(hand)-1):
            if hand[i][1] > hand[i+1][1]:
                hand[i],hand[i+1] = hand[i+1],hand[i]
            else:
                if hand[i][1] == hand[i+1][1]:
                    if hand[i][1] == 'red' and hand[i+1][1] != 'red':
                        hand[i],hand[i+1] = hand[i+1],hand[i]
                    if hand[i][1] == 'green' and hand[i+1][1] == 'blue':
                        hand[i],hand[i+1] = hand[i+1],hand[i]
                
    return hand

print("Hand 1 before sorting: ",hand1)
print('------------------------------')
print("Hand 1 after sorting ",sort(hand1))
print('------------------------------')
print("Hand 2 before sorting: ",hand2)
print('------------------------------')
print("Hand 1 after sorting ",sort(hand2))
```

```python
#Task 1.4

def calc_score(hand1,hand2):
    score_1 =0 
    score_2 = 0
    for i in range(len(hand1)):
        winner_card =compare_colours(hand1[i],hand2[i])
        if winner_card == 1 :
            score_1 += hand1[i][1] * 2
        if winner_card == 2:
            score_2 += hand2[i][1] * 2
        if winner_card == 3:
            score_1 += hand1[i][1]
            score_2 += hand2[i][1] 
            
    return score_1,score_2

print(calc_score(hand1,hand2))
        
```

```python
#Task 1.5

score = calc_score(hand1,hand2)
hand1_score = score[0]
hand2_score = score[1]

if hand1_score > hand2_score:
    print("The winner is player 1")
    print("The second place goes to player 2")
else:
    print("The winner is player 2")
    print("The second place goes to player 1")
```
