```python
#2024 A Level Paper 2
#Task 2

#Task 2.1

class Node:
    def __init__(self):
        self.data_value = -1
        self.left_node = -1
        self.right_node = -1

    def set_data(self,value):
        self.data_value = value

    def set_left(self,left):
        self.left_node = left

    def set_right(self,right):
        self.right_node = right

    def get_data(self):
        return self.data_value

    def get_left(self):
        return self.left_node

    def get_right(self):
        return self.right_node
```

```python

#Task 2.2

def calculate_nodes(no_of_levels):
    if no_of_levels != 0:
        nodes = 1
        for i in range(1,no_of_levels):
            nodes += (2 ** (i))

        display = ''
        display += "A "
        display += str(no_of_levels)
        display += " level number tree has "
        display += str(nodes)
        display += " nodes"
        print(display)
        return nodes
```

```python
#Task 2.3
print(calculate_nodes(1))
print(calculate_nodes(3))
print(calculate_nodes(5))
print(calculate_nodes(10))
```

```python
#Task 2.4

def create_empty_tree(list_of_nodes,no_of_levels):
    nodes = calculate_nodes(no_of_levels)
    
    for i in range(nodes):
        node = Node()
        list_of_nodes.append(node)

    return list_of_nodes
```

```python

#Task 2.5

def assign_leaf_nodes(list_of_nodes,no_of_levels):
    no_of_leaf_nodes = 2 ** (no_of_levels-1)
    count = 1
    for i in range(no_of_leaf_nodes):
        node = list_of_nodes[i]
        node.set_data(count)
        count += 1

    return list_of_nodes
```

```python
#Task 2.6

def create_number_tree(list_of_nodes,no_of_levels):
    total_nodes = calculate_nodes(no_of_levels)
    start_count = 0
    count = 2 ** (no_of_levels-1)
    no_of_levels = no_of_levels-1
    
    while no_of_levels != 0:
        j = 0
        for i in range(start_count,(count),2):
            left_subtree = list_of_nodes[i]

            right_subtree = list_of_nodes[i+1]

            data = left_subtree.get_data() + right_subtree.get_data()
            
            
            list_of_nodes[count+j].set_data(data)
            list_of_nodes[count+j].set_left(i)
            list_of_nodes[count+j].set_right(i+1)

            j+=1
            
        temp = count
        start_count = count
        count = temp + (2 ** (no_of_levels-1))
        
        no_of_levels = no_of_levels-1
        

    return list_of_nodes
```

```python

#Task 2.7
lst_of_nodes = create_empty_tree([],3)

lst_of_nodes = assign_leaf_nodes(lst_of_nodes,3) 
lst_of_nodes = create_number_tree(lst_of_nodes,3)

print("Left Pointer"+"\t"+"Data","\t","Right Pointer")

for node in lst_of_nodes:
    print(node.get_left(),"\t\t",node.get_data(),"\t\t",node.get_right())





```
