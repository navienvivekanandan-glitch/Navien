```python
import sqlite3,datetime
#Task 4.2

class Person:
    def __init__(self,name,dob):
        self.full_name = name
        self.date_of_birth = dob

    def is_adult(self):
        now = datetime.datetime.now()
        curr_year = now.year
        curr_month = now.month
        curr_day = now.day
        dob_year,dob_month,dob_day = self.date_of_birth.split('-')

        if int(curr_year) - int(dob_year) >= 18:
            if int(curr_month) < int(dob_month):
                return False
            if int(curr_month) == int(dob_month) and int(curr_day) < int(dob_day):
                return False
            else:
                return True
        else:
            return False

    def screen_name(self):
        screen_name = ''
        for i in range(len(self.full_name)):
            if self.full_name[i].isalpha() == True:
                screen_name += self.full_name[i]

        dob_year,dob_month,dob_day = self.date_of_birth.split('-')
        screen_name += dob_month
        screen_name += dob_day

        return screen_name

class Staff(Person):
    def __init__(self,name,dob):
        super().__init__(name,dob)

    def is_adult(self):
        return True

    def screen_name(self):
        return self.full_name + "Staff"

class Student(Person):
    def __init__(self,name,dob):
        super().__init__(name,dob)

    def is_adult(self):
        return False



file = open('people.txt','r')
for line in file:
    line = line.strip().split(',')
    name,DOB,person_type = line[0],line[1],line[2]

    conn = sqlite3.connect('school.db')
    if person_type == "Person":
        person = Person(name,DOB)
        FullName = name
        DateOfBirth = DOB
        ScreenName = person.screen_name()
        if person.is_adult() == True:
            IsAdult = 1
        else:
            IsAdult = 0
    elif person_type == "Student":
        student = Student(name,DOB)
        FullName = name
        DateOfBirth = DOB
        ScreenName = student.screen_name()
        IsAdult = 0
    else:
        staff = Staff(name,DOB)
        FullName = name
        DateOfBirth = DOB
        ScreenName = staff.screen_name()
        IsAdult = 1


    conn.execute('''INSERT INTO People(FullName,DateOfBirth,ScreenName,IsAdult)
    VALUES (?,?,?,?);''',(FullName,DateOfBirth,ScreenName,IsAdult))

    conn.commit()

conn.close()
file.close()
```
