```python
#Task 3.1

def task3_1(quantity_of_data):
    n = len(quantity_of_data) - 2
    number = quantity_of_data[:n]
    string = quantity_of_data[n:]
    
    if string == "KB":
        value = int(number) * (10**3)
    elif string == "MB":
        value = int(number) * (10**6)
    elif string == "GB":
        value = int(number) * (10**9)
    elif string == "TB":
        value = int(number) * (10**12)
    else:
        return "invalid data"
        
    return value


print(task3_1("8KB"))
```

```python
#Task 3.2

def task3_2(quantity_of_data):
    if task3_1(quantity_of_data) != "invalid data":
        return task3_1(quantity_of_data)
    n = len(quantity_of_data) - 3
    number = quantity_of_data[:n]
    string = quantity_of_data[n:]
    if string == "KiB":
        value = int(number) * (2**10)
    elif string == "MiB":
        value = int(number) * (2**20)
    elif string == "GiB":
        value = int(number) * (2**30)
    elif string == "TiB":
        value = int(number) * (2**40)
    else:
        return "invalid data"
    return value

print(task3_2("2MiB"))
```

```python
#Task 3.3

def task3_3(quantity_of_data,target_unit):
    test = ['KB','MB','GB','TB','KiB','MiB','GiB','TiB']

    if target_unit not in test:
        return "invalid data"
    else:
        byte = task3_2(quantity_of_data)

        if target_unit == 'KB':
            value = byte / (10**3)
        elif target_unit == 'MB':
            value = byte / (10**6)
        elif target_unit == 'GB':
            value = byte / (10**9)
        elif target_unit == 'TB':
            value = byte / (10**12)
        elif target_unit == 'KiB':
            value = byte / (2**10)
        elif target_unit == 'MiB':
            value = byte / (2**20)
        elif target_unit == 'GiB':
            value = byte / (2**30)
        elif target_unit == 'TiB':
            value = byte / (2**40)
        return value

print(task3_3("512MiB","GiB"))
```
