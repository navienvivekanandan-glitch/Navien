```python
#Task 2.1

class QueueItem:
    def __init__(self,ID,quantity):
        self.id = ID
        self.quantity = quantity
        
    def getID(self):
        return self.id
    
    def getQuantity(self):
        return self.quantity

#Task 2.2

class Queue:
    def __init__(self):
        self.queueList = [None]*4
        self.headPointer = 0
        self.tailPointer = 0
        self.numberItems = 0
        
#Task 2.3    

    def enqueue(self,QueueItem):
        if self.tailPointer == 0 and self.headPointer == 0:
            self.queueList[self.tailPointer] = QueueItem
            self.numberItems += 1
            self.tailPointer += 1
            return True
        else:
            if self.numberItems < 4:
                self.queueList[self.tailPointer] = QueueItem
                self.numberItems += 1
                self.tailPointer += 1
                return True
            else:
                return False

#Task 2.4

    def dequeue(self):
        if self.tailPointer == 0:
            return 'Queue is empty'
        else:
            queue_item = self.queueList[self.headPointer]
            self.headPointer += 1
            self.numberItems -= 1
            if self.headPointer >= self.tailPointer:
                self.headPointer = 0
                self.tailPointer = 0
            ID = queue_item.getID()
            quantity = queue_item.getQuantity()

            return (ID,quantity)

#Task 2.5

    def printQueue(self):
        for i in range(self.headPointer,self.tailPointer):
            queue_item = self.queueList[i]
            ID = queue_item.getID()
            quantity = queue_item.getQuantity()
            print("ID: ",ID,"Quantity: ",quantity)
            
            
```

```python
#Task 2.3
queue = Queue()

for i in range(5):
    ID = input("Enter the ID: ")
    Quantity = int(input("Enter the Quantity: "))
    
    item = QueueItem(ID,Quantity)
    
    print(queue.enqueue(item))
```
