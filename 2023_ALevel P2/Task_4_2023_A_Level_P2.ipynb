```python
#Task 4

#Task 4.1


def read_file(filename):
    lst = []
    file = open(filename,'r')
    for line in file:
        line = line.strip()
        lst.append(line)

    file.close()
    return lst

def compress_data(lst):
    string = ''
    count = 0
    for i in range(len(lst)-1):
        while lst[i] == lst[i+1]:
            count += 1
        else:
            compress = lst[i] + str(count)
            print(compress)
            string += compress
            print(string)
            count = 0

    return string

def write_file(string):
    file = open('compressedimage.txt','w')
    file.write(string)
    file.close()
    return

```
