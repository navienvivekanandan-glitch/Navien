```python
#Task 1.1
import sqlite3

conn = sqlite3.connect('TESTDATABASE')


conn.execute('DROP TABLE IF EXISTS Student')
conn.execute('DROP TABLE IF EXISTS test')
conn.execute('DROP TABLE IF EXISTS Class')
conn.execute('DROP TABLE IF EXISTS Student_Test')


conn.execute('''CREATE TABLE IF NOT EXISTS `Student` (
	`StudentID`	INTEGER NOT NULL UNIQUE,
	`GivenName`	TEXT NOT NULL,
	`FamilyName`	TEXT NOT NULL,
	`ClassName`	TEXT NOT NULL
);''')

conn.execute('''CREATE TABLE IF NOT EXISTS `Test` (
	`TestID`	TEXT NOT NULL UNIQUE,
	`MaxMarks`	INTEGER NOT NULL
);''')

conn.execute('''CREATE TABLE IF NOT EXISTS `Class` (
	`ClassName`	TEXT NOT NULL UNIQUE,
	`YearGroup`	INTEGER
);''')

conn.execute('''CREATE TABLE `Student_Test` (
	`StudentTestID`	INTEGER NOT NULL UNIQUE,
	`StudentID`	INTEGER NOT NULL UNIQUE,
	`TestID`	TEXT NOT NULL UNIQUE,
	`Mark`	INTEGER NOT NULL,
	`DateTest`	TEXT NOT NULL,
	FOREIGN KEY(`StudentID`) REFERENCES `Student`(`StudentID`)
);''')

conn.commit()
conn.close()
```
