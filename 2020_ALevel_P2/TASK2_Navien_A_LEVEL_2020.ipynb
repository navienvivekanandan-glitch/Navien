```python
#Task 2.1
import random
def task2_1(filename,quantity,maximum):
    file = open(filename,'w')
    for i in range(quantity):
        number = random.randint(0,maximum)
        file.write(str(number) + '\n')

    file.close()


task2_1('randomnumbers.txt',1000,5000)
```

```python
#Task 2.2

def task2_2(list_of_integers):
    n = len(list_of_integers)

    if n == 1:
        return list_of_integers

    mid = n //2

    list1 = list_of_integers[:mid]
    list2 = list_of_integers[mid:]

    list1 = task2_2(list1)
    list2 = task2_2(list2)

    return merge(list1,list2)


def merge(a,b):
    c = []
    i = 0
    j = 0

    while i < len(a) and j < len(b):
        if a[i] > b[j]:
            c.append(b[j])
            j += 1
        else:
            c.append(a[i])
            i += 1

    while i < len(a):
        c.append(a[i])
        i += 1

    while j < len(b):
        c.append(b[j])
        j += 1

    return c

list_of_integers = [56,25,4,98,0,18,4,5,7,0]

print(task2_2(list_of_integers))
```

```python
#Task 2.3

def task2_3(filename_in,filename_out):
    file = open(filename_in,'r')
    lst = []
    for line in file:
        line = line.strip()
        lst.append(int(line))

    file.close()

    sorted_list = task2_2(lst)

    file = open(filename_out,'w')
    for i in range(len(sorted_list)):
        file.write(str(sorted_list[i]) + '\n')
        
    file.close()


task2_3('randomnumbers.txt','sorted_numbers.txt')
```
